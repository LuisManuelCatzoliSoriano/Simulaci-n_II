```python
from random import random
from statistics import mean, variance

def f(n):
  l=[]
  ll=[]
  for i in range(n):
    u=random()
    x=1/(1+u)
    x1=1/(1+(1-u))
    l.append(x)
    ll.append((x+x1)/2)
  return l,ll

n=100000
l,ll=f(n)
m=mean(l)
m1=mean(ll)
v=variance(l)
v1=variance(ll)
print(v,v1,100*(v-v1)/v)
```
