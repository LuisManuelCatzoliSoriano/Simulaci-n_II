```python
import numpy as np
```

```python
def f(delt,T):
  sig=np.sqrt(delt)
  t=0
  x=1
  s=0
  while t<T:
    W=np.random.normal(0,sig,1)
    s=s+W
    x=x+delt+2*np.sqrt(x-1)*W
    t=t+delt
  z=1+s**2
  return (x-z)**2
```

```python
T=0.2
delt=0.1
f(delt,T)
```

Segunda ecuación

```python
def f2(delt,T):
    sig=np.sqrt(delt)
    t=0
    x=10
    s=0

    while t<T:
        W=sig*np.random.normal(0,1)

        x=x+0.5*x*delt+0.1*x*W
        s=s+W

        t=t+delt

    z=10*np.exp((0.5-(0.1**2/2))*t+0.1*s)

    return (x-z)**2
```

```python
T=0.2
delt=0.1
f2(delt,T)
```

```python
def simu(n):
    l=[]
    for i in range(n):
        p=f2(0.1,0.2)
        l.append(p)
    prom=sum(l)/n
    return prom
```

```python
simu(1000)
```
