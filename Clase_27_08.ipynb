Estimar por simulación de Montecarlo $\sqrt{2}$

```python
def g(y):
  return np.sqrt(2)
```

```python
def experimento_n(M):
  I=0
  for i in range(M):
    x=r.random()
    I+=g(x)
  return I/M
```

```python
experimento_n(1000)
```
