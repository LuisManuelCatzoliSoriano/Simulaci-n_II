Estimar por simulación de Monte Carlo el coeficiente de correlación entre $W_{10}$ y $W_{20}$.

La idea es:

$$W_{10}\sim N(0,10)$$

$$W_{20}\sim N(0,20)$$

y como:

$$W_{20}=W_{10}+(W_{20}-W_{10})$$

```python
import numpy as np
from scipy.stats import pearsonr
```

```python
N = 100000

W10 = np.sqrt(10) * np.random.randn(N)

W20_W10 = np.sqrt(10) * np.random.randn(N)

W20 = W10 + W20_W10

correlacion, pvalor = pearsonr(W10, W20)

print("Correlación estimada:", correlacion)
print("Valor teórico:", np.sqrt(10/20))
print("p-valor:", pvalor)
```

Dibujar 4 trayectorias para $W_0,W_1,W_2,...,W_{100}$

```python
import matplotlib.pyplot as plt
```

```python
n_trayectorias = 4

n_pasos = 100

incrementos = np.random.randn(n_trayectorias, n_pasos)

W = np.zeros((n_trayectorias, n_pasos + 1))

W[:, 1:] = np.cumsum(incrementos, axis=1)

t = np.arange(0, n_pasos + 1)

for i in range(n_trayectorias):
    plt.plot(t, W[i, :], label=f"Trayectoria {i+1}")

plt.xlabel("t")
plt.ylabel("$W_t$")
plt.title("4 trayectorias del movimiento browniano")
plt.legend()
plt.grid(True)
plt.show()
```
