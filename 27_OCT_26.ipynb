```python
from random  import gauss
import numpy as np
from numba import jit,njit,prange
@njit(parallel=True,fastmath=True)
def f(x0,c,u,o,delta,T,ns):
    oo=np.sqrt(delta)
    n=int(T/delta)
    l=np.zeros(ns)
    ll=np.zeros(ns)
    for i in prange(ns):
        s1=x0
        s2=x0
        for j in prange(n):
            w=oo*gauss(0,1)
            s1=s1+c*(s1-u)*delta+o*w
            s2=s2+c*(s2-u)*delta-o*w
        l[i]=s1
        ll[i]=(s1+s2)/2
    return l,ll
c=-0.4
u=7
o=0.2
x0=7
T=2
ns=1000000
delta=0.1
l,ll=f(x0,c,u,o,delta,T,ns)
v=np.var(l)
v1=np.var(ll)
print(v,v1,100*(v-v1)/v)
m=np.mean(l)
m1=np.mean(ll)
print(m,m1)
for i in range(10):
    print("%10.6f"%l[i], "%10.6f"%ll[i])
```

```python
from random  import gauss, random
import numpy as np
from numba import jit,njit,prange
from scipy.stats import pearsonr
@njit(nopython=True)
def f(x):
    return 1/(1+x)
@jit(nopython=True)
def control(x):
    return -x
@njit(parallel=True,fastmath=True)
def ff(ns):
    lx=np.zeros(ns)
    ly=np.zeros(ns)
    lz=np.zeros(ns)
    for i in range(ns):
        u=random()
        x=f(u)
        y=control(u)
        lx[i]=x
        ly[i]=y
    return lx,ly
ns=100000
lx,ly=ff(ns)
print(pearsonr(lx,ly)[0]**2)
co=np.cov(lx,ly)
print(co)
c=-co[0,1]/co[1,1]
print(c)
```
